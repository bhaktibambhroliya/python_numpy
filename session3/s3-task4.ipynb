```python
import numpy as np

playlist = np.array([
    [10, 5, 8],
    [7, 12, 6],
    [15, 4, 9]
])

print("Original Playlist Grid:", playlist)

transpose_T = playlist.T
print("Transpose using T:", transpose_T)

transpose_func = np.transpose(playlist)
print("Transpose using np.transpose():", transpose_func)
```
