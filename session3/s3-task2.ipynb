```python
import numpy as np

steps = np.array([
    [2000, 3000],
    [2500, 3500],
    [3000, 4000],
    [3500, 4500],
    [4000, 5000]
])

print("original 2D array:", steps)
print("shape of the array:", steps.shape)

steps_1D = steps.reshape(10)
print("reshaped 1D array:", steps_1D)
print("shape of the reshaped array:", steps_1D.shape)

steps_2D = steps_1D.reshape(5, 2)
print("reshaped back to 2D array:", steps_2D)
print("shape of the reshaped array:", steps_2D.shape)




```
