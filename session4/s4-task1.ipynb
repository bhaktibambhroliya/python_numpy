```python
import numpy as np

likes = np.array([120, 250, 340, 180, 500, 420, 300])

comments = np.array([15, 25, 30, 20, 45, 35, 28])

engagement = likes + comments

average_engagement = np.sum(engagement) / len(engagement)

print("Likes:", likes)
print("Comments:", comments)
print("Total Engagement:", engagement)
print("Average Engagement per Post:", average_engagement)
```
