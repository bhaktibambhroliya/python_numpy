```python
import numpy as np

np.random.seed(42)
ratings = np.random.uniform(1, 5, 10)

rounded_ratings = np.round(ratings)

floor_ratings = np.floor(ratings)

ceil_ratings = np.ceil(ratings)

print("Original Ratings:", ratings)
print("Rounded Ratings:", rounded_ratings)
print("Floor Ratings:", floor_ratings)
print("Ceil Ratings:", ceil_ratings)
```
