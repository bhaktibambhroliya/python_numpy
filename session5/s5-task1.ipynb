```python
import numpy as np

my_likes = np.array([120, 250, 340, 180, 500, 420, 300])

friend_likes = np.array([100, 200, 300, 220, 450, 400, 350])

combined_likes = np.add(my_likes, friend_likes)

difference = np.subtract(my_likes, friend_likes)

# Print the results
print("My Likes:", my_likes)
print("Friend's Likes:", friend_likes)
print("Combined Likes:", combined_likes)
print("Difference in Likes:", difference)
```
